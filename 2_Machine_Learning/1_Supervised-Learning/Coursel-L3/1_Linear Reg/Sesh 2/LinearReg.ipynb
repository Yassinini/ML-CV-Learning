# Linear Regression

## Import libraries/Data

```python
import numpy as np
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt 
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

## Clean Data

```python
d = pd.read_csv('melon.csv')
d
```

```python
d.info()
d.isna().sum()
```

## Visualise

```python
corr = d.corr()
plt.figure(figsize=(10, 8))
sns.scatterplot(d, x = d['Weight'], y = d['Price'])
plt.title('Weight vs Price')
plt.xlabel('Weight')
plt.ylabel('Price')
```

## Train_Test

### Identify Values

```python
x=d['Weight'].values.reshape(-1,1)
y=d['Price'].values.reshape(-1, 1)
print("x: ", x)
print("y: ", y)
```

### Split Data

```python
x_train, x_test, y_train, y_test = train_test_split(x , y , train_size = 0.7)
print("x_train: ", x_train)
print("x_test: ", x_test)
print("y_train: ", y_train)
print("y_test: ", y_test)
```

## Train LINEAR REGRESSION Model

```python
model = LinearRegression()
model.fit(x_train, y_train)
pred = model.predict(x_test)
```

### Predicted values and R2 score 

```python
print("Predicted values: ", pred)
print("R2 Score: ", r2_score(y_test, pred))
```

```python
pred.dtype
```

## Present Pred vs Actual on a graph

```python
plt.scatter(x_test, y_test, color='blue', label='Actual')
plt.scatter(x_test, pred, color='red', label='Predicted')
plt.plot(x_test, y_test, color='blue', label='Actual', alpha=0.5)
plt.plot(x_test, pred, color='red', label='Predicted', alpha=0.5)
plt.legend()
plt.title('Actual vs Predicted')
plt.xlabel('Weight')
plt.ylabel('Price')
plt.show()
```
