```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.preprocessing import LabelEncoder
```

```python
d = pd.read_csv("final_health_project.csv")
d.head()
```

```python
d.shape
```

```python
d.drop('ID',axis=1,inplace=True)
```

```python
d.isna().sum()
d.fillna(d.mean(numeric_only=True), inplace=True)
d['Diet_Quality'] = d['Diet_Quality'].fillna(d['Diet_Quality'].mode()[0])
d.isna().sum()
```

```python
d.head(10)
```

```python
d.duplicated().sum()
d.drop_duplicates(inplace=True)
d.duplicated().sum()
```

```python
le = LabelEncoder()
d['Diet_Quality'] = le.fit_transform(d['Diet_Quality'])
d['BMI_Category'] = le.fit_transform(d['BMI_Category'])
d
```

```python
x = d.drop('BMI_Category',axis=1)
y = d['BMI_Category']
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 42)
```

```python
lg = LogisticRegression(max_iter=1000)
lg.fit(x_train, y_train)
pred = lg.predict(x_test)
print(accuracy_score(y_test, pred))
```
