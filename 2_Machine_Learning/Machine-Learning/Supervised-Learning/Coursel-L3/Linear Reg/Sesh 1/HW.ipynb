## Import Libraries

```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
```

## Filter data

```python
d = pd.read_csv('attachment_45928_Titanic-Dataset.csv.csv')
d.head(10)
```

```python
d.info()
```

```python
d.isna().sum()
d.dropna(inplace=True)
d.info
```

```python
d.shape
```

## Make heatmap matrix

```python
le = LabelEncoder()
d['Sex'] = le.fit_transform(d['Sex'])
```

```python
corr = d.corr(numeric_only=True)
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()
```
