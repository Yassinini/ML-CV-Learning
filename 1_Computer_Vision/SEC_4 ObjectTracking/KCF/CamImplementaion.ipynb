## KCF Camera Implementation

## Import cv2

```python
import cv2 as cv
vid = cv.VideoCapture(0)
```

## Preprocess frame and set condition

```python
ret, frame = vid.read()
if not ret:
    print("I CANT SEEEEEEEEEEEEEEEE")
    vid.release()
    exit()
frame = cv.resize(frame, (800,600))
frame = cv.flip(frame,1)
```

```python
tracker= cv.TrackerKCF_create()
bbox= cv.selectROI("image", frame)
cv.destroyWindow("image")
tracker.init(frame, bbox)
```

##  Run Video

```python
while True:
    if not ret:
        break
    ret, frame = vid.read()
    image= cv.cvtColor= (frame, cv.COLOR_BGR2RGB)
    frame = cv.resize(frame, (800,600))
    frame = cv.flip(frame,1)

    success, bbox = tracker.update(frame)
    if success:
        x, y, w, h = [int(v) for v in bbox]
        cv.rectangle(frame, (x,y), (x+w, y+h), (255,0,0), 2)
    else:
        cv.putText(frame, "Error fr cant see nun" , (400,300), cv.FONT_HERSHEY_SIMPLEX, 1, (255,0,0), 2)

    cv.imshow("image", frame)
    if cv.waitKey(1) & 0xFF == 27:
        break
vid.release()
cv.destroyAllWindows()
```
