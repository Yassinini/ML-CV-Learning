# Tracking USAIN BOLT in a clip of a race

## import library

```python
import cv2 as cv
```

## Initialise Tracker and Video

```python
from pathlib import Path

# Get repo root for file paths
repo_root = Path.cwd()
while repo_root.name != 'Machine_Learning' and repo_root.parent != repo_root:
    repo_root = repo_root.parent
course_materials = repo_root / 'computer_vision' / 'Course_Materials'

tracker =cv.TrackerKCF_create()
vid = cv.VideoCapture(str(course_materials / 'Videos' / 'race.mp4'))
ok, frame = vid.read()
```

```python
boundbox= cv.selectROI(frame)
ok= tracker.init(frame, boundbox)
```

## Run Video

```python
while True:
    ok , frame= vid.read()
    if not ok:
        break
    ok, boundbox = tracker.update(frame)
    if ok:
        x,y,w,h = [int(v) for v in boundbox]
        cv.rectangle(frame, (x,y), (x+w, y+h), (0,240,140), 2)
    else:
        cv.putText(frame, "Done i think", (100,80), cv.FONT_HERSHEY_SIMPLEX, 1, (255,0,0), 2)
    cv.imshow("Video", frame)
    if cv.waitKey(1) & 0XFF ==27:
        break
vid.release()
cv.destroyAllWindows()
```
