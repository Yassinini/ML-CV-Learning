```python
import os
import cv2 as cv
import dlib
from PIL import Image
import numpy as np
```

```python
from pathlib import Path

notebook_dir = Path('/home/suupatruupa/Documents/@Code (Copy)/Machine_Learning/computer_vision/Face Recognition/DLIB FR')
weights_dir = notebook_dir.parent.parent / 'Course_Materials' / 'Weights'
datasets_dir = notebook_dir.parent / 'samples' / 'Datasets' / 'yalefaces' / 'train'

frontal_face_detector=dlib.get_frontal_face_detector()
shape_predictor=dlib.shape_predictor(str(weights_dir / 'shape_predictor_68_face_landmarks.dat'))
face_recognition_model=dlib.face_recognition_model_v1(str(weights_dir / 'dlib_face_recognition_resnet_model_v1.dat'))
index={}
current_index=0
face_descriptors=None
paths=[os.path.join(str(datasets_dir), f) for f in os.listdir(str(datasets_dir))]
```

```python
for path in paths:
    img = Image.open(path).convert('RGB')
    image_array=np.array(img, 'uint8')
    detected_faces=frontal_face_detector(image_array,1)
    for face in detected_faces:
        left , top , right , bottom = face.left(), face.top(), face.right(), face.bottom()
        cv.rectangle(image_array, (left,top), (right,bottom), (100,0,200),2)
        points=shape_predictor(image_array, face)
        for point in points.parts():
            cv.circle(image_array, (point.x, point.y), 2 , (0,255,0), 1)
        current_face_descriptor=face_recognition_model.compute_face_descriptor(image_array, points)
        current_face_descriptor=[f for f in current_face_descriptor]
        current_face_descriptor=np.array(current_face_descriptor, np.float64)
```

```python
print(current_face_descriptor)
```

```python
current_face_descriptor= current_face_descriptor[np.newaxis, :]
print(current_face_descriptor.shape) # (1,128)
print(current_face_descriptor) #matrices
```

```python
if face_descriptors is None:
    face_descriptors = current_face_descriptor
else:
    face_descriptors = np.concatenate((face_descriptors, current_face_descriptor), axis =0)
index[current_index]=path
current_index+=1
print(face_descriptors.shape)
```
